```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = "https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/cereal.csv"
df = pd.read_csv(url)
df.head()
```

```python
fields = ['shelf','weight','cups','rating']
cereal_df_new = df.drop(fields, axis=1)
cereal_df_new.head()
```

```python
#corr()

cereal_corr = cereal_df_new.corr(numeric_only=True)
cereal_corr
```

```python
#ones_like tạo một mảng mà các giá trị chỉ có T F và size nó giống cái size của data
ones_like = np.ones_like(cereal_corr, dtype=bool)
ones_like
```

```python
#np's triu: trả về giá trị true ở đường chéo trên
mask = np.triu(ones_like)
mask
```

```python
sns.heatmap(data = cereal_corr, mask=mask)
```

```python
mask_new = mask[1:, :-1]
mask_new
```

```python
cereal_corr_new = cereal_corr.iloc[1:, :-1]
cereal_corr_new
```

```python
sns.heatmap(data = cereal_corr_new, mask = mask_new, cmap = "Greens", linecolor = "white", linewidths= 0.5)
```
