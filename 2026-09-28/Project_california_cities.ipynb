```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
plt.style.available
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
cities = pd.read_csv('C:/Users/Admin/Learning_AI/daily-learning-AI/2026-09-28/california_cities.csv')
cities.head()

#cities['area_total_km2'].sort_values(ascending = False)
```

```python
# Tách 2 colums là latd(vi do) và longtd(kinh do)

lat, long = cities['latd'], cities['longd']
lat, long
```

```python

population, area = cities['population_total'], cities['area_total_km2']
plt.scatter(long, lat, c = np.log10(population), cmap = "viridis", s = area);
plt.xlabel("longtitude");
plt.ylabel("latitude");
plt.colorbar(label = 'log$_(10)$(population)');
area_range = [50, 100, 200, 500]
for i in area_range:
    plt.scatter([], [], s = i, label = str(i) + 'km^2', c = 'k', alpha = 0.5);
plt.legend()
```
