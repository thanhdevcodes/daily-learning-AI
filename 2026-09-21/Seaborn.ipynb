## Distribution Plot
### Univariate distribution of observations

```python
tips_df = sns.load_dataset('tips')
```

```python
tips_df.head()
```

```python
sns.histplot(data = tips_df["total_bill"])
```

# KDE

```python
sns.kdeplot(data = tips_df["total_bill"])
```

# Bar plot

```python
tips_df.head()
```

```python
sns.barplot(data = tips_df, x = "sex", y = "tip", estimator= np.mean)
```

# count plot

```python
tips_df["sex"].value_counts()
```

```python
sns.countplot(data = tips_df, x = "sex")
```

# Boxlot

```python
sns.boxplot(data = tips_df, x = "day", y = "total_bill", hue = "sex");
```

# Facet grid

```python
tips_fg = sns.FacetGrid(data = tips_df, row = "smoker", col = "time")
tips_fg.map(sns.scatterplot, "total_bill",  "tip")
```

```python
tips_df.head()
```

```python
kws = dict(s = 100, edgecolor = 'b', alpha = .7)
new_fg = sns.FacetGrid(data = tips_df,col = "sex", col_order=["Female","Male"], hue = "smoker")
new_fg.map(sns.scatterplot, "total_bill", "tip", **kws)

new_fg.add_legend()
```

# Joint plot

```python
new_df = sns.load_dataset('penguins')
new_df.head()
```

```python
sns.jointplot(data = new_df, x = "flipper_length_mm", y = "bill_length_mm", hue = "species");
```

# pair plot

```python
sns.pairplot(data = new_df, hue = "species");
```

# Heatmaps

```python
fl_df = sns.load_dataset("flights")
fl_df.head()
```

```python
flights = pd.pivot_table(fl_df, index="month", columns="year", values="passengers")
flights
```

```python
sns.heatmap(data = flights, cmap = "Greens")
```
