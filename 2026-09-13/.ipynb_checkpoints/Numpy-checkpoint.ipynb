#  Dot Product example

```python
sales_amounts = np.random.randint(1, 20, (4, 3))
sales_amounts
```

```python
import pandas as pd
```

```python
week_sales = pd.DataFrame(sales_amounts, index = ["Thứ 2", "Thứ 3", "Thứ 4", "Thứ 5"],
                                        columns = ["Nước", "Thịt", "Rau"])
week_sales
```

```python
coins = np.random.randint(1, 15, (3, 1))
coins
```

```python
coins_item = pd.DataFrame(coins, index = ["Nước", "Thịt", "Rau"], columns = ["Giá"])
coins_item
```

```python
total = sales_amounts @ coins
total
```

```python
tongdoanhthu = pd.DataFrame(total, index = ["Thứ 2", "Thứ 3", "Thứ 4", "Thứ 5"], columns = ["Giá"]) ### cach 1
tongdoanhthu
```

```python
tongdoanhthu1 = week_sales.dot(coins_item) ## cach 2
tongdoanhthu1
```

```python
week_sales["Giá"] = tongdoanhthu1
week_sales
```
