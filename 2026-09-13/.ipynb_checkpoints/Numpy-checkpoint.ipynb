### tạo một cái mảng numpy từ các hàm có sẵn

```python
np.zeros((4, 5))
```

```python
np.zeros((4, 5), dtype = int)
```

```python
np.zeros(2) ### tạo 1 mảng chỉ toàn số 0
```

```python
np.ones((2, 6), dtype = int) ### tạo một mảng mà chỉ toàn số 1
```

```python
np.arange(0, 100, 4) ### tạo một mảng có điểm đầu và điểm cuối có bước nhảy 
```

```python
np.full((2, 9), 67) ### tạo một mảng cùng một giá trị nhập vào 
```

```python
np.linspace(0, 3, 7, dtype = int) ### in ra các giá trị từ đầu đến cuối theo số giá trị đã cho 
```

```python
np.random.normal(0, 1, (3, 3))
```

```python
a3 = np.random.randint(4, 9, size = 6)
print(a3)
```

```python
a3 = np.random.randint(4, 9, (3, 4))
print(a3)
```

### ARRAY INDEXING AND SLICING

### one- dimensional subarray

```python
x1 = np.random.randint(4, 10, 8)
print(x1)
```

```python
print(x1[-1], x1[5], x1[0])
```

### multi- dimensional subarray

```python
x2 = np.random.randint(1, 10, (3, 4))
x2
```

```python
print(x2[2, 3])
```

```python
x2[2, 3] = 10
```

```python
x2
```

### Slicing
### x[start:end:step]

```python
x1
```

```python
x1[3:8]
```

```python
x1[::4]
```

```python
x2
```

```python
x2[:2, :3]
```

```python
x2[1:2, 0:2]
```

```python
x2[:2]
```

```python
x2[:, :3]
```
