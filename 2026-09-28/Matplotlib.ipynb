### Pyplot API vs Object-Oriented(OO) API
* Pyplot APT -> Nhanh, dễ
* OO -> Khó hơn

### Pyplot API

```python
x = np.linspace(0, 10, 1000)
plt.plot(x, np.sin(x), linestyle = "dashed", label = "sin(x)");
plt.plot(x, np.cos(x), linestyle = "dashed", color = "red", label = "cos(x)");
plt.title("sin(x) and cos(x)")
plt.xlabel("Bien")
plt.ylabel("sin(x)")

#plt.xlim(0, 4)
#plt.ylim(0, 1)
#plt.axis([0, 4, 0, 1])
#plt.axis("tight")
plt.axis("equal")

plt.legend()
```

### OO API

```python
#1.prepare data
x = [1, 2, 3, 4]
y = [11, 22, 33, 44]

#2.setup plot
fig, ax = plt.subplots(figsize = (5, 5))

#3. plot the data
ax.plot(x, y);
ax.set(title = "A simple plot", xlabel = "x", ylabel = "y");
```

```python
##OO API

#Prepare new data
new = np.random.RandomState(0)
x = new.randn(100)
y = new.randn(100)

colors = new.rand(100)
sizes = 1000*new.rand(100)

fig, ax = plt.subplots()
img1 = ax.scatter(x, y, s = sizes, c = colors, cmap = 'viridis');

fig.colorbar(img1);
```

## 3. Bar: biểu đồ cột

```python
#Prepare data
soft_drinks_data = {"Coca": 10, "Pepsi": 15, "7up": 12}

fig, ax = plt.subplots()
ax.bar(soft_drinks_data.keys(), soft_drinks_data.values());
ax.set(title = "Sieu thi", xlabel = "Drinks", ylabel = "Cost")
```

## 4.Histogram: biểu đồ theo tần suất

```python
#Prepare data
np.random.seed(42)
students_height = np.random.normal(170, 10, 250)

students_height[:6]
```

```python
plt.hist(students_height);
```

```python
fig, ax = plt.subplots()
ax.hist(students_height, bins= 100);
```

### Subplots
https://matplotlib.org/3.1.1/gallery/recipes/create_subplots.html

```python
#Option 1:
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows= 2, ncols= 2, figsize = (10, 5));
ax1.plot(x, np.sin(x), color = "blue");
ax1.plot(x, x/2, color = "red");
ax2.scatter(np.random.rand(10), np.random.rand(10), s = 100*np.random.rand(10), c = np.random.rand(10));
ax3.bar(soft_drinks_data.keys(), soft_drinks_data.values());
ax4.hist(students_height, bins= 100);
```

```python
#Option 2:

fig, ax = plt.subplots(nrows= 2, ncols= 2, figsize = (10, 5));
ax[0, 0].plot(x, np.sin(x), color = "blue");
ax[0, 0].plot(x, x/2, color = "red");
ax[0, 1].scatter(np.random.rand(10), np.random.rand(10), s = 100*np.random.rand(10), c = np.random.rand(10));
ax[1, 0].bar(soft_drinks_data.keys(), soft_drinks_data.values());
ax[1, 1].hist(students_height, bins= 100);
```
